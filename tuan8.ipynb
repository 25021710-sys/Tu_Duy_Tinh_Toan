```python
class Calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2

    def subtract(self):
        return self.num1 - self.num2

    def product(self):
        return self.num1 * self.num2

    def divide(self):

        if self.num2 == 0:
            raise ValueError("Không thể chia cho 0!")
        return self.num1 / self.num2

    def power(self):
        
        return self.num1 ** self.num2

    def mod(self):
        
        if self.num2 == 0:
            raise ValueError("Không thể chia dư cho 0!")
        return self.num1 % self.num2


if __name__ == "__main__":
    while True:

        print("\nMenu:")
        print("1. add (Cộng)")
        print("2. subtract (Trừ)")
        print("3. product (Nhân)")
        print("4. divide (Chia)")
        print("5. power (Lũy thừa)")
        print("6. mod (Chia dư)")
        try:
            choice = int(input("Chọn phép tính (1-6): "))
            if choice < 1 or choice > 6:
                print("Lựa chọn không hợp lệ! Vui lòng chọn từ 1 đến 6.")
                continue
            num1 = float(input("Nhập số thứ nhất: "))
            num2 = float(input("Nhập số thứ hai: "))
            calc = Calculator(num1, num2)
            if choice == 1:
                result = calc.add()
                print(f"Kết quả cộng: {result}")
            elif choice == 2:
                result = calc.subtract()
                print(f"Kết quả trừ: {result}")
            elif choice == 3:
                result = calc.product()
                print(f"Kết quả nhân: {result}")
            elif choice == 4:
                result = calc.divide()
                print(f"Kết quả chia: {result}")
            elif choice == 5:
                result = calc.power()
                print(f"Kết quả lũy thừa: {result}")
            elif choice == 6:
                result = calc.mod()
                print(f"Kết quả chia dư: {result}")
        
        except ValueError as e:
            
            print(f"Lỗi: {e}")
            continue
         continue_choice = input("Bạn có muốn tiếp tục? (YES/Yes để tiếp tục): ")
        if continue_choice.lower() != "yes":
            print("Kết thúc chương trình.")
            break
    
```

```python
class ShoppingCart:
    def __init__(self):
        self.cart = []

    def add_product(self, name, price):
        self.cart.append({'name': name, 'price': price})

    def remove_product(self, name):
        self.cart = [product for product in self.cart if product['name'] != name]

    def is_empty(self):
        return len(self.cart) == 0

    def total_price(self):
        return sum(product['price'] for product in self.cart)

    def show_products(self):
        if self.is_empty():
            print("Giỏ hàng đang trống.")
        else:
            print("Các sản phẩm trong giỏ hàng:")
            for product in self.cart:
                print(f"- {product['name']}: {product['price']} VND")

    def clear_cart(self):
        self.cart.clear()


def menu():
    cart = ShoppingCart()
    while True:
        print("\nChọn chức năng:")
        print("2. Thêm sản phẩm")
        print("3. Xóa sản phẩm")
        print("4. Kiểm tra giỏ hàng có rỗng không")
        print("5. Tính tổng tiền")
        print("6. Hiển thị sản phẩm")
        print("7. Xóa toàn bộ giỏ hàng")
        choice = input("Nhập lựa chọn (2-7): ")

        if choice == "2":
            name = input("Nhập tên sản phẩm: ")
            price = float(input("Nhập giá sản phẩm: "))
            cart.add_product(name, price)
        elif choice == "3":
            name = input("Nhập tên sản phẩm cần xóa: ")
            cart.remove_product(name)
        elif choice == "4":
            print("Giỏ hàng rỗng." if cart.is_empty() else "Giỏ hàng có sản phẩm.")
        elif choice == "5":
            print(f"Tổng tiền: {cart.total_price()} VND")
        elif choice == "6":
            cart.show_products()
        elif choice == "7":
            cart.clear_cart()
            print("Đã xóa toàn bộ giỏ hàng.")
        else:
            print("Lựa chọn không hợp lệ.")

        exit_program = input("Có muốn thoát khỏi chương trình không? (yes/no): ")
        if exit_program.lower() == "yes":
            print("Chương trình kết thúc.")
            break

menu()
```

```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

    def perimeter(self):
        return 2 * (self.w + self.h)

    def scale(self, k):
        self.w *= k
        self.h *= k

w, h, k = map(float, input("Nhập w, h, k cách nhau bởi dấu cách: ").split())
rect = Rectangle(w, h)
print(f"{rect.area():.2f} {rect.perimeter():.2f}")
rect.scale(k)
print(f"{rect.w:.2f} {rect.h:.2f}")
```

```python
import math

class Fraction:
    def __init__(self, num, den):
        if den == 0:
            raise ValueError("Mẫu số không được bằng 0")
        if den < 0:
            num, den = -num, -den
        gcd = math.gcd(num, den)
        self.num = num // gcd
        self.den = den // gcd

    def add(self, other):
        return Fraction(self.num * other.den + other.num * self.den, self.den * other.den)

    def sub(self, other):
        return Fraction(self.num * other.den - other.num * self.den, self.den * other.den)

    def mul(self, other):
        return Fraction(self.num * other.num, self.den * other.den)

    def div(self, other):
        return Fraction(self.num * other.den, self.den * other.num)

    def __str__(self):
        return f"{self.num}/{self.den}"
-1

a, b, op, c, d = input().split()
a, b, c, d = int(a), int(b), int(c), int(d)
f1 = Fraction(a, b)
f2 = Fraction(c, d)

if op == '+':
    result = f1.add(f2)
elif op == '-':
    result = f1.sub(f2)
elif op == '*':
    result = f1.mul(f2)
elif op == '/':
    result = f1.div(f2)

print(result)
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = float(balance)

    def deposit(self, x):
        self.balance += float(x)

    def withdraw(self, x):
        if self.balance >= float(x):
            self.balance -= float(x)

    def __str__(self):
        return f"{self.owner}: {self.balance:.2f}"

owner, balance = input().split()
account = BankAccount(owner, balance)
n = int(input())

for _ in range(n):
    cmd, x = input().split()
    if cmd == "DEPOSIT":
        account.deposit(x)
    elif cmd == "WITHDRAW":
        account.withdraw(x)

print(f"{account.balance:.2f}")
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

    def add_score(self, subject, score):
        self.scores.append(score)

    def avg(self):
        return sum(self.scores) / len(self.scores)

    def rank(self):
        a = self.avg()
        if a >= 8:
            return "Excellent"
        elif a >= 6.5:
            return "Good"
        elif a >= 5:
            return "Average"
        else:
            return "Poor"

name = input().strip()
n = int(input())

st = Student(name)

for _ in range(n):
    subject, score = input().split()
    st.add_score(subject, float(score))

average = st.avg()
print(f"{st.name} {average:.2f} {st.rank()}")
```

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year


class Library:
    def __init__(self):
        self.books = []

    def add(self, book):
        self.books.append(book)

    def count_by_author(self, author):
        return sum(1 for b in self.books if b.author == author)

    def find_by_year(self, year):
        return sum(1 for b in self.books if b.year == year)


n = int(input())
lib = Library()

for _ in range(n):
    line = input().strip()

    if line.startswith("ADD"):
        _, data = line.split(" ", 1)
        title, author, year = data.split(";")
        lib.add(Book(title, author, int(year)))

    elif line.startswith("COUNTYEAR"):
        _, year = line.split()
        print(lib.find_by_year(int(year)))

    elif line.startswith("COUNT"):
        _, author = line.split()
        print(lib.count_by_author(author))
```

```python
import math

class Point:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z

    def vector(self, other):
        return (
            other.x - self.x,
            other.y - self.y,
            other.z - self.z
        )

def cross(u, v):
    return (
        u[1]*v[2] - u[2]*v[1],
        u[2]*v[0] - u[0]*v[2],
        u[0]*v[1] - u[1]*v[0]
    )

def dot(u, v):
    return u[0]*v[0] + u[1]*v[1] + u[2]*v[2]

def norm(v):
    return math.sqrt(dot(v, v))

A = Point(*map(float, input().split()))
B = Point(*map(float, input().split()))
C = Point(*map(float, input().split()))
D = Point(*map(float, input().split()))

AB = A.vector(B)
AC = A.vector(C)
BC = B.vector(C)
BD = B.vector(D)

n1 = cross(AB, AC)
n2 = cross(BC, BD)

cos_theta = abs(dot(n1, n2)) / (norm(n1) * norm(n2))
angle = math.degrees(math.acos(cos_theta))

print(f"{angle:.2f}")
```

```python
class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def add(self, other):
        return ComplexNumber(
            self.real + other.real,
            self.imag + other.imag
        )

    def multiply(self, other):
        return ComplexNumber(
            self.real * other.real - self.imag * other.imag,
            self.real * other.imag + self.imag * other.real
        )

a, b = map(float, input().split())
c, d = map(float, input().split())

z1 = ComplexNumber(a, b)
z2 = ComplexNumber(c, d)

sum_z = z1.add(z2)
mul_z = z1.multiply(z2)

print(f"{int(sum_z.real)} {int(sum_z.imag)}")
print(f"{int(mul_z.real)} {int(mul_z.imag)}")
```
