```python
def find(s,n):
    low,high= 0,len(s)-1
    while low<=high:
        mid =(low+high)//2
        if s[mid]==n:
            return mid
        elif n>s[mid]:
            low= mid+1
        else: high=mid-1
    return -1
s= list(map(int,input().split()))
n=int(input())
print(find(s,n))
```

```python
def length_of_LIS(nums):
    if not nums:
        return 0
    
    n = len(nums)
    dp = [1] * n
    
    for i in range(1, n):
        for j in range(i):
            if nums[j] < nums[i]:
                dp[i] = max(dp[i], dp[j] + 1)
    
    return max(dp)
nums = list(map(int, input().split()))
print(length_of_LIS(nums))
```

```python
import bisect 
def length_of_LIS(nums):
    if not nums:
        return 0

    tails=[]
    for num in nums:
        if not tails or num>tails[-1]:
            tails.append(num)
        else:
            k=bisect.bisect_left(tails,num)
            tails[k]=num
    return len(tails)

nums = list(map(int, input().split()))
print(length_of_LIS(nums))
```

```python
def shortest_covering_interval(intervals, queries):
    result = []
    for q in queries:
        min_length = float('inf')
        found=False
        for l, r in intervals:
            if l <= q <= r:
                length = r - l + 1
                if length < min_length:
                    min_length = length
                found = True
        if found:
            result.append(min_length)
        else:
            result.append(-1)
    return result

intervals = [[1,4], [2,4], [3,6], [4,4]]
queries = [2,3,4,5]
print(shortest_covering_interval(intervals, queries)) 
```
