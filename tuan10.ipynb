```python
class InvalidDivisionError(Exception):
    pass

# Nhập hai số thực a, b
a, b = map(float, input().split())

if b == 0:
    raise InvalidDivisionError("Lỗi chia cho không")
print("{:.2f}".format(a / b))
```

```python
def is_happy(n):
    seen = set()
    while n != 1 and n not in seen:
        seen.add(n)
        n = sum(int(digit) ** 2 for digit in str(n))
    return n == 1

# Nhập số nguyên dương a
a = int(input())

if is_happy(a):
    print("YES")
else:
    print("NO")
```

```python
def is_abundant(n):
    if n < 1:
        return False
    sum_div = 1
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            sum_div += i
            if i != n // i:
                sum_div += n // i
    return sum_div > n

# Nhập số nguyên dương a
a = int(input())

if is_abundant(a):
    print("YES")
else:
    print("NO")
```

```python
import math

class InvalidDivisionError(Exception):
    pass

# Nhập số thực a
a = float(input())

if a < 0:
    raise InvalidDivisionError("Loai chia khong")
print("{:.2f}".format(math.sqrt(a)))
```

```python
class InvalidAgeError(Exception):
    pass

try:
    N = int(input())
    if N <= 0:
        raise InvalidAgeError("Tuổi không hợp lệ")
    print(2025-N)
except ValueError:
    raise InvalidAgeError("Tuổi không hợp lệ")
```
